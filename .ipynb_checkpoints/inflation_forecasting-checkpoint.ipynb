# Inflation Forecasting Using Food and Energy Price Indicators
**Time Series Forecasting Project**

This notebook predicts the Consumer Price Index (CPI) using macroeconomic indicators such as food prices, crude oil prices, fuel/energy prices, and commodity indices.

The primary focus is **India**, configured via the `COUNTRY_CONFIG` block.


## Section 0: Setup & Configuration
Import libraries and set up the global configuration for the notebook.


```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import wbdata
import pandas_datareader.data as web
import datetime
import warnings

from statsmodels.tsa.stattools import adfuller, kpss, grangercausalitytests
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.tsa.vector_ar.var_model import VAR
from statsmodels.tools.eval_measures import rmse, mse
import pmdarima as pm

from sklearn.metrics import mean_absolute_error, mean_absolute_percentage_error, r2_score
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.model_selection import train_test_split
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings('ignore')
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams['figure.figsize'] = (14, 6)

# ─── COUNTRY CONFIGURATION ────────────────────────────────────────────────────
COUNTRY_CONFIG = {
    "India":         {"wb_code": "IN", "cpi_indicator": "FP.CPI.TOTL", "oil_series": "DCOILBRENTEU"},
    "United States": {"wb_code": "US", "cpi_indicator": "FP.CPI.TOTL", "oil_series": "DCOILWTICO"},
    "Brazil":        {"wb_code": "BR", "cpi_indicator": "FP.CPI.TOTL", "oil_series": "DCOILBRENTEU"},
    "Germany":       {"wb_code": "DE", "cpi_indicator": "FP.CPI.TOTL", "oil_series": "DCOILBRENTEU"},
    "China":         {"wb_code": "CN", "cpi_indicator": "FP.CPI.TOTL", "oil_series": "DCOILBRENTEU"},
}
ACTIVE_COUNTRY = "India"
CFG = COUNTRY_CONFIG[ACTIVE_COUNTRY]

print(f"Project configured for: {ACTIVE_COUNTRY}")
```

## Section 1: Dataset Sourcing & Loading
Fetch data programmatically from World Bank and FRED.

**Indicators:**
1. CPI (Target) - World Bank (`FP.CPI.TOTL`)
2. Crude Oil - FRED (Brent for India, WTI for US)
3. Food Price Index - FRED (`PFOODINDEXM`)
4. Energy Index - FRED (`PNRGINDEXM`)
5. Agricultural Raw Materials - FRED (`PRAWMINDEXM`)
6. Metals & Minerals - FRED (`PMETINDEXM`)


```
# 1. Fetch CPI from World Bank
start_date = datetime.datetime(2000, 1, 1)
end_date = datetime.datetime(2024, 12, 1)

countries = [CFG["wb_code"]]
# Note: World Bank data through wbdata v0.3+ requires tuple format for indicators or direct fetching
indicator = {CFG["cpi_indicator"]: "CPI"}

print(f"Fetching {ACTIVE_COUNTRY} CPI data from World Bank...")
try:
    cpi_df = wbdata.get_dataframe(indicator, country=countries, data_date=(start_date, end_date))
    cpi_df.index = pd.to_datetime(cpi_df.index)
    cpi_df = cpi_df.sort_index()
    # Interpolate from annual to monthly if World Bank only returns annual
    cpi_monthly = cpi_df.resample('MS').interpolate('linear')
    print("World Bank CPI fetched successfully.")
except Exception as e:
    print(f"Error fetching WB data: {e}")
    # Fallback to FRED's Global Price Index of All Commodities as a proxy if WB fails
    cpi_monthly = web.DataReader('PALLFNFINDEXM', 'fred', start_date, end_date)
    cpi_monthly.columns = ['CPI']
    print("Using FRED global proxy instead.")

# 2. Fetch Macroeconomic Indicators from FRED
fred_series = {
    CFG["oil_series"]: 'Oil_Price',
    'PFOODINDEXM': 'Food_Index',
    'PNRGINDEXM': 'Energy_Index',
    'PRAWMINDEXM': 'Agri_Index',
    'PMETINDEXM': 'Metals_Index'
}

print("Fetching exogenous variables from FRED...")
fred_df = web.DataReader(list(fred_series.keys()), 'fred', start_date, end_date)
fred_df.rename(columns=fred_series, inplace=True)

# 3. Merge Datasets
merged_df = pd.merge(cpi_monthly, fred_df, left_index=True, right_index=True, how='inner')

# Forward fill missing values then drop remaining
merged_df = merged_df.ffill().dropna()

print(f"Dataset Shape: {merged_df.shape}")
display(merged_df.head())
```

## Section 2: Exploratory Data Analysis (EDA)


```
# 1. Time Series Plots
fig, axes = plt.subplots(nrows=3, ncols=2, figsize=(18, 12))
plt.subplots_adjust(hspace=0.4)

axes[0,0].plot(merged_df.index, merged_df['CPI'], color='blue')
axes[0,0].set_title(f"Target: {ACTIVE_COUNTRY} CPI")

axes[0,1].plot(merged_df.index, merged_df['Oil_Price'], color='black')
axes[0,1].set_title("Crude Oil Price")

axes[1,0].plot(merged_df.index, merged_df['Food_Index'], color='green')
axes[1,0].set_title("Food Price Index")

axes[1,1].plot(merged_df.index, merged_df['Energy_Index'], color='orange')
axes[1,1].set_title("Energy Price Index")

axes[2,0].plot(merged_df.index, merged_df['Agri_Index'], color='brown')
axes[2,0].set_title("Agricultural Raw Materials Index")

axes[2,1].plot(merged_df.index, merged_df['Metals_Index'], color='purple')
axes[2,1].set_title("Metals Index")

plt.show()
```

```
# 2. Correlation Heatmap
plt.figure(figsize=(8,6))
sns.heatmap(merged_df.corr(), annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title("Correlation between Indicators and CPI")
plt.show()
```

```
# 3. Stationarity Tests (ADF Test)
def test_stationarity(series, title):
    result = adfuller(series.dropna())
    print(f"--- {title} Stationarity ---")
    print(f"ADF Statistic: {result[0]:.4f}")
    print(f"p-value: {result[1]:.4f}")
    if result[1] <= 0.05:
        print("=> Data is Stationary (Reject Null Hypothesis)")
    else:
        print("=> Data is Non-Stationary (Fail to Reject Null Hypothesis)")
    print("")

for col in merged_df.columns:
    test_stationarity(merged_df[col], col)
```

```
# 4. Seasonal Decomposition
decomp = seasonal_decompose(merged_df['CPI'].dropna(), model='additive', period=12)
fig = decomp.plot()
fig.set_size_inches(14, 8)
plt.show()
```

```
# 5. ACF and PACF for CPI
fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(16,4))
plot_acf(merged_df['CPI'].dropna(), ax=ax[0], lags=40, title='ACF - CPI')
plot_pacf(merged_df['CPI'].dropna(), ax=ax[1], lags=40, title='PACF - CPI')
plt.show()

# Since CPI is generally non-stationary, let's look at 1st diff
cpi_diff = merged_df['CPI'].diff().dropna()
fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(16,4))
plot_acf(cpi_diff, ax=ax[0], lags=40, title='ACF - Differenced CPI')
plot_pacf(cpi_diff, ax=ax[1], lags=40, title='PACF - Differenced CPI')
plt.show()
```

## Section 3: Data Preprocessing
- First order differencing for stationarity
- Lag feature engineering
- Train/Test Split


```
# 1. Differencing
df_diff = merged_df.diff().dropna()

print("Testing stationarity on differenced target...")
test_stationarity(df_diff['CPI'], "Differenced CPI")

# 2. Lag Feature Engineering (for ML models)
df_ml = merged_df.copy()

lags = [1, 3, 6]
exog_cols = ['Oil_Price', 'Food_Index', 'Energy_Index', 'Agri_Index', 'Metals_Index']

for col in exog_cols + ['CPI']:
    for lag in lags:
        df_ml[f"{col}_lag{lag}"] = df_ml[col].shift(lag)

df_ml = df_ml.dropna()

print(f"Dataset for ML Shape: {df_ml.shape}")

# 3. Train/Test Split (80/20 Chronological)
train_size = int(len(merged_df) * 0.8)
train, test = merged_df.iloc[:train_size], merged_df.iloc[train_size:]

train_diff, test_diff = df_diff.iloc[:train_size-1], df_diff.iloc[train_size-1:]

ml_train_size = int(len(df_ml) * 0.8)
ml_train, ml_test = df_ml.iloc[:ml_train_size], df_ml.iloc[ml_train_size:]

print(f"Train dates: {train.index.min().date()} to {train.index.max().date()}")
print(f"Test dates:  {test.index.min().date()} to {test.index.max().date()}")
```

## Section 4: ARIMA / SARIMA (Univariate Baseline)

```
# Auto ARIMA
print("Fitting Auto ARIMA on CPI...")
ar_model = pm.auto_arima(train['CPI'], 
                      start_p=1, start_q=1,
                      test='adf',       
                      max_p=3, max_q=3, 
                      m=12,              
                      start_P=0, seasonal=True,   
                      d=1, D=1, trace=True,
                      error_action='ignore',  
                      suppress_warnings=True, 
                      stepwise=True)

print(ar_model.summary())

# Forecast
n_periods = len(test)
sarima_pred, sarima_confint = ar_model.predict(n_periods=n_periods, return_conf_int=True)
sarima_pred.index = test.index

# Metrics
sarima_rmse = rmse(test['CPI'], sarima_pred)
sarima_mae = mean_absolute_error(test['CPI'], sarima_pred)
sarima_mape = mean_absolute_percentage_error(test['CPI'], sarima_pred)
sarima_r2 = r2_score(test['CPI'], sarima_pred)

print(f"\nSARIMA RMSE: {sarima_rmse:.4f}")
print(f"SARIMA MAPE: {sarima_mape:.4f}")

plt.plot(train.index, train['CPI'], label='Train CPI')
plt.plot(test.index, test['CPI'], label='Actual Test CPI')
plt.plot(test.index, sarima_pred, label='SARIMA Forecast', color='red')
plt.fill_between(test.index, sarima_confint[:, 0], sarima_confint[:, 1], color='pink', alpha=0.3)
plt.legend()
plt.title("SARIMA Forecast vs Actual")
plt.show()
```

## Section 5: Vector Autoregression (VAR)

```
# VAR Model on differenced data to avoid spurious regression
model_var = VAR(train_diff)
sorted_order = model_var.select_order(maxlags=12)
print(sorted_order.summary())

# Using lag=3 (example) based on AIC
var_result = model_var.fit(3)

# Forecast
lag_order = var_result.k_ar
var_forecast = var_result.forecast(train_diff.values[-lag_order:], steps=len(test_diff))
var_pred_diff = pd.DataFrame(var_forecast, index=test_diff.index, columns=train_diff.columns)

# Undifference to get back to levels (starts from last train value)
last_train_cpi = train['CPI'].iloc[-1]
var_pred_levels = var_pred_diff['CPI'].cumsum() + last_train_cpi
var_pred_levels = var_pred_levels[test.index] # Align index

var_rmse = rmse(test['CPI'], var_pred_levels)
var_mae = mean_absolute_error(test['CPI'], var_pred_levels)
var_mape = mean_absolute_percentage_error(test['CPI'], var_pred_levels)
var_r2 = r2_score(test['CPI'], var_pred_levels)

print(f"VAR RMSE: {var_rmse:.4f}")

plt.plot(train.index[-24:], train['CPI'].iloc[-24:], label='Train (last 2 years)')
plt.plot(test.index, test['CPI'], label='Actual Test')
plt.plot(test.index, var_pred_levels, label='VAR Forecast', color='purple')
plt.legend()
plt.title("VAR Forecast vs Actual")
plt.show()
```

## Section 6: SARIMAX (ARIMA with Exogenous Variables)

```
# Using Food, Oil, and Energy as exogenous
exog_train = train[['Food_Index', 'Oil_Price', 'Energy_Index']]
exog_test = test[['Food_Index', 'Oil_Price', 'Energy_Index']]

# Fit SARIMAX using order found by auto_arima
order = ar_model.order
seasonal_order = ar_model.seasonal_order

print(f"Fitting SARIMAX{order}{seasonal_order} with Exogenous...")
sarimax = SARIMAX(train['CPI'], exog=exog_train, order=order, seasonal_order=seasonal_order)
sarimax_result = sarimax.fit(disp=False)

# Forecast
sarimax_pred = sarimax_result.predict(start=test.index[0], end=test.index[-1], exog=exog_test)

sarimax_rmse = rmse(test['CPI'], sarimax_pred)
sarimax_mae = mean_absolute_error(test['CPI'], sarimax_pred)
sarimax_mape = mean_absolute_percentage_error(test['CPI'], sarimax_pred)
sarimax_r2 = r2_score(test['CPI'], sarimax_pred)

print(f"SARIMAX RMSE: {sarimax_rmse:.4f}")

plt.plot(test.index, test['CPI'], label='Actual Test')
plt.plot(test.index, sarimax_pred, label='SARIMAX Forecast', color='orange')
plt.legend()
plt.title("SARIMAX Forecast vs Actual")
plt.show()
```

## Section 7: XGBoost Model

```
# Features and Target
X_train = ml_train.drop(columns=['CPI'] + exog_cols) # Only use lag features
y_train = ml_train['CPI']
X_test = ml_test.drop(columns=['CPI'] + exog_cols)
y_test = ml_test['CPI']

xgb_model = XGBRegressor(n_estimators=100, learning_rate=0.05, max_depth=4, random_state=42)
xgb_model.fit(X_train, y_train)

xgb_pred = pd.Series(xgb_model.predict(X_test), index=X_test.index)

xgb_rmse = rmse(y_test, xgb_pred)
xgb_mae = mean_absolute_error(y_test, xgb_pred)
xgb_mape = mean_absolute_percentage_error(y_test, xgb_pred)
xgb_r2 = r2_score(y_test, xgb_pred)

print(f"XGBoost RMSE: {xgb_rmse:.4f}")

# Feature Importance
importances = pd.Series(xgb_model.feature_importances_, index=X_train.columns).sort_values(ascending=True)
importances.tail(10).plot(kind='barh', color='teal', figsize=(10,5))
plt.title("XGBoost Top 10 Feature Importances")
plt.show()

lgbm_model = LGBMRegressor(n_estimators=100, learning_rate=0.05, max_depth=4, random_state=42, verbose=-1)
lgbm_model.fit(X_train, y_train)
lgbm_pred = pd.Series(lgbm_model.predict(X_test), index=X_test.index)

lgbm_rmse = rmse(y_test, lgbm_pred)
lgbm_mae = mean_absolute_error(y_test, lgbm_pred)
lgbm_mape = mean_absolute_percentage_error(y_test, lgbm_pred)
lgbm_r2 = r2_score(y_test, lgbm_pred)
```

## Section 8: LSTM (Deep Learning)

```
scaler_X = MinMaxScaler()
scaler_y = MinMaxScaler()

scaled_X_train = scaler_X.fit_transform(X_train)
scaled_X_test = scaler_X.transform(X_test)

scaled_y_train = scaler_y.fit_transform(y_train.values.reshape(-1, 1))
scaled_y_test = scaler_y.transform(y_test.values.reshape(-1, 1))

# Reshape for LSTM: [samples, time steps, features]
# Here we treat time steps=1 and features=all lagged columns
X_train_lstm = scaled_X_train.reshape((scaled_X_train.shape[0], 1, scaled_X_train.shape[1]))
X_test_lstm = scaled_X_test.reshape((scaled_X_test.shape[0], 1, scaled_X_test.shape[1]))

lstm = Sequential([
    LSTM(64, activation='relu', input_shape=(X_train_lstm.shape[1], X_train_lstm.shape[2]), return_sequences=True),
    Dropout(0.2),
    LSTM(32, activation='relu'),
    Dense(1)
])

lstm.compile(optimizer='adam', loss='mse')

early_stop = EarlyStopping(monitor='val_loss', patience=15, restore_best_weights=True)

history = lstm.fit(X_train_lstm, scaled_y_train, epochs=100, batch_size=16, 
                   validation_split=0.1, callbacks=[early_stop], verbose=0)

# Predict and Inverse Transform
lstm_pred_scaled = lstm.predict(X_test_lstm)
lstm_pred = scaler_y.inverse_transform(lstm_pred_scaled).flatten()

lstm_rmse = rmse(y_test, lstm_pred)
lstm_mae = mean_absolute_error(y_test, lstm_pred)
lstm_mape = mean_absolute_percentage_error(y_test, lstm_pred)
lstm_r2 = r2_score(y_test, lstm_pred)

print(f"LSTM RMSE: {lstm_rmse:.4f}")
```

## Section 9: Model Evaluation & Comparison

```
results_df = pd.DataFrame({
    'Model': ['SARIMA', 'VAR', 'SARIMAX', 'XGBoost', 'LightGBM', 'LSTM'],
    'RMSE': [sarima_rmse, var_rmse, sarimax_rmse, xgb_rmse, lgbm_rmse, lstm_rmse],
    'MAE': [sarima_mae, var_mae, sarimax_mae, xgb_mae, lgbm_mae, lstm_mae],
    'MAPE': [sarima_mape, var_mape, sarimax_mape, xgb_mape, lgbm_mape, lstm_mape],
    'R2_Score': [sarima_r2, var_r2, sarimax_r2, xgb_r2, lgbm_r2, lstm_r2]
})

results_df = results_df.sort_values('RMSE').reset_index(drop=True)
display(results_df)

# Plot overall comparison
plt.figure()
sns.barplot(data=results_df, x='Model', y='RMSE', palette='viridis')
plt.title("RMSE Comparison Across Models")
plt.show()
```

## Section 10: Forecasting 12-Months Ahead

```
# Using the baseline SARIMA to forecast the unknown future
# Jan 2025 to Dec 2025
future_dates = pd.date_range(start='2025-01-01', periods=12, freq='MS')

future_pred, future_conf = ar_model.predict(n_periods=len(test) + 12, return_conf_int=True)
future_pred = future_pred.iloc[-12:]
future_conf = future_conf[-12:]

plt.plot(train.index[-36:], train['CPI'].iloc[-36:], label='Train History')
plt.plot(test.index, test['CPI'], label='Test History')
plt.plot(future_dates, future_pred, label='SARIMA 12-Month Forecast', color='red')
plt.fill_between(future_dates, future_conf[:, 0], future_conf[:, 1], color='pink', alpha=0.3)
plt.title("12-Month Forward CPI Forecast")
plt.legend()
plt.show()

print("Forecast Values (Jan 2025 - Dec 2025):")
display(pd.DataFrame({'Forecast': future_pred.values}, index=future_dates))
```

## Section 11: Insights & Granger Causality

```
# 1. Granger Causality Tests
# Test if 'Exogenous' causes 'CPI'
print("Granger Causality Test Results (Do Macro Indicators cause CPI?)\n")

max_lag = 3
for col in exog_cols:
    print(f"--- Does {col} cause CPI? ---")
    data_gc = df_diff[['CPI', col]].dropna()
    try:
        res = grangercausalitytests(data_gc, maxlag=[max_lag], verbose=False)
        p_value = res[max_lag][0]['ssr_ftest'][1]
        print(f"p-value (lag {max_lag}): {p_value:.4f}")
        if p_value < 0.05:
            print("=> YES: Strong evidence of causality")
        else:
            print("=> NO: Insufficient evidence of causality")
    except Exception as e:
        print("Test failed.")
    print("\n")
```
